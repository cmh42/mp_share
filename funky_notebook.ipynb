```python
def funky_rabbit(number): 
    for _ in range(number):
        print("Funky Rabbit")
    return None
```

```python
def whacky_fox(max_distance):
    steps = 0
    while steps < max_distance: 
        print("run", end=" ")
        steps += 1
        if steps % 8 == 0: 
            print()
        
    return None 
```

```python
whacky_fox(20)
```
