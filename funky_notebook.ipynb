```python
x_values = np.linspace(-6*np.pi , 6*np.pi, 800)
trig_fun = lambda x: 3 * (np.sin(x))**2 - 5*np.cos(x)
plt.plot(x_values, trig_fun(x_values), label="Special function")
plt.legend()
plt.ylim(-8, 8)
plt.title("An interesting function")
```

```python
def funky_rabbit(number): 
    for _ in range(number):
        print("Funky Rabbit")
    return None
```

```python
funky_rabbit(8)
```

```python
def whacky_fox(max_distance):
    steps = 0
    while steps < max_distance: 
        print("run", end=" ")
        steps += 1
        if steps % 8 == 0: 
            print()
        
    return None 
```

```python
whacky_fox(20)
```
